```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout
from sklearn.metrics import confusion_matrix, classification_report
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt

# Setting up data generators with augmented data for training
train_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=30,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=20,  # Reduced shear range
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'  # Added fill mode
)

validation_datagen = ImageDataGenerator(rescale=1./255)
test_datagen = ImageDataGenerator(rescale=1./255)

# Loading data from the newly organized folders
train_generator = train_datagen.flow_from_directory(
    'archive/Face Mask Dataset/Train',  # Changed to forward slashes
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary'
)

validation_generator = validation_datagen.flow_from_directory(
    'archive/Face Mask Dataset/Validation',  # Changed to forward slashes
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary'
)

test_generator = test_datagen.flow_from_directory(
    'archive/Face Mask Dataset/Test',  # Changed to forward slashes
    target_size=(150, 150),
    batch_size=32,
    class_mode='binary',
    shuffle=False
)

# Building the model
model = Sequential([
    Conv2D(32, (3, 3), activation='relu', input_shape=(150, 150, 3)),
    MaxPooling2D(2, 2),
    Conv2D(64, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Conv2D(128, (3, 3), activation='relu'),
    MaxPooling2D(2, 2),
    Flatten(),
    Dense(128, activation='relu'),
    Dropout(0.5),  # Increased dropout rate
    Dense(1, activation='sigmoid')
])

# Compiling the model
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])

# Training the model
history = model.fit(
    train_generator,
    epochs=15,  # Increased epochs for better training
    validation_data=validation_generator
)

# Saving the model after training
model.save('mask_detection_model.h5')

# Evaluating the model on the test set
test_loss, test_accuracy = model.evaluate(test_generator)
print(f'Test accuracy: {test_accuracy * 100:.2f}%')

# Generating predictions on the test set
y_pred = model.predict(test_generator)
y_pred_classes = (y_pred > 0.5).astype("int32")
y_true = test_generator.classes

# Confusion Matrix and Classification Report
cm = confusion_matrix(y_true, y_pred_classes)
print("Confusion Matrix:\n", cm)
print("Classification Report:\n", classification_report(y_true, y_pred_classes))

# Plotting training history
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.plot(history.history['accuracy'], label='Training Accuracy')
plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
plt.title('Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(history.history['loss'], label='Training Loss')
plt.plot(history.history['val_loss'], label='Validation Loss')
plt.title('Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.show()

# Loading a new image and predicting
def predict_image(image_path):
    try:
        img = Image.open(image_path).convert('RGB')  # Convert to RGB
        img = img.resize((150, 150))  # Resize to match model input
        img_array = np.array(img) / 255.0
        img_array = np.expand_dims(img_array, axis=0)
        predictions = model.predict(img_array)
        
        print("Prediction probabilities:", predictions)  # Print probabilities
        
        # Adjusted threshold
        threshold = 0.5
        if predictions[0][0] < threshold:
            print("The person is wearing a mask")
        else:
            print("The person is not wearing a mask")

        # Display the image
        img.show()
    except Exception as e:
        print(f"Error loading or predicting image: {e}")

# Example usage of the function
# predict_image('path_to_your_image.jpg')
```

```python
import cv2
import numpy as np
from tensorflow.keras.models import load_model

# Load the trained model
model = load_model('mask_detection_model.h5')

# Start video capture from the camera
cap = cv2.VideoCapture(0)

# Load the face detector (Haar Cascade)
face_cascade_path = 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(face_cascade_path)

# Check if the Haar Cascade file is loaded properly
if face_cascade.empty():
    raise IOError(f"Error loading Haar Cascade classifier from {face_cascade_path}")

while True:
    # Read a frame from the camera
    ret, frame = cap.read()

    if not ret:
        print("Error: Unable to capture frame from video stream")
        break

    # Convert the image to grayscale for face detection
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Check if the frame is valid
    if gray is None:
        print("Error: Invalid frame")
        continue

    # Detect faces in the image
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=5)

    if len(faces) == 0:
        print("No faces detected")

    for (x, y, w, h) in faces:
        # Crop the face region from the image
        face_roi = frame[y:y+h, x:x+w]

        # Prepare the image for the model
        img = cv2.resize(face_roi, (150, 150))
        img = img / 255.0
        img = np.expand_dims(img, axis=0)

        # Predict using the model
        prediction = model.predict(img)

        # Determine the label based on the prediction
        label = "Mask detected" if prediction[0][0] < 0.001 else "No mask detected"
        color = (0, 255, 0) if prediction[0][0] < 0.001 else (0, 0, 255)

        # Draw a rectangle around the face
        cv2.rectangle(frame, (x, y), (x+w, y+h), color, 2)

        # Determine the position of the label to appear above the rectangle
        label_position = (x, y - 10) if y - 10 > 10 else (x, y + h + 30)

        # Write the label on the image
        cv2.putText(frame, label, label_position, cv2.FONT_HERSHEY_SIMPLEX, 0.9, color, 2)

    # Display the resulting frame
    cv2.imshow('Face Mask Detection', frame)

    # Check for the 'q' key press or the close window event
    if cv2.waitKey(1) & 0xFF == ord('q') or cv2.getWindowProperty('Face Mask Detection', cv2.WND_PROP_VISIBLE) < 1:
        break

# Release the video capture object and close all OpenCV windows
cap.release()
cv2.destroyAllWindows()
```
